```python
## Importing Data Set,
## I intend to calculate the average energy effeciency score
## for buildings registered by the NYC Department of Buildings
```

```python
## importing data set
import pandas as pd

law33 = pd.read_csv("DOB_Sustainability_Compliance_Map__Local_Law_33_20261112.csv")

law33.head()

law33.dtypes
```

```python
## Mean,Median,Mode

Energy_Score_Mean = law33["ENERGY STAR Score"].mean()
print(Energy_Score_Mean)

Energy_Score_Median = law33["ENERGY STAR Score"].median()
print(Energy_Score_Median)

Energy_Score_Mode = law33["ENERGY STAR Score"].mode()
print(Energy_Score_Mode)
```

```python
## Creating a visualization for median score per borough

MAN = law33[law33["BoroughName"] == "MANHATTAN"]
BK = law33[law33["BoroughName"] == "BROOKLYN"]
QNS = law33[law33["BoroughName"] == "QUEENS"]
BX = law33[law33["BoroughName"] == "BRONX"]
SI = law33[law33["BoroughName"] == "STATEN ISLAND"]

MAN_mean = MAN["ENERGY STAR Score"].mean()
print(MAN_mean)

BK_mean = BK["ENERGY STAR Score"].mean()
print(BK_mean)

QNS_mean = QNS["ENERGY STAR Score"].mean()
print(QNS_mean)

BX_mean = BX["ENERGY STAR Score"].mean()
print(BX_mean)

SI_mean = SI["ENERGY STAR Score"].mean()
print(SI_mean)

borough_means_dict = {
    "MANHATTAN": MAN_mean,
    "BROOKLYN": BK_mean,
    "QUEENS": QNS_mean,
    "BRONX": BX_mean,
    "STATEN ISLAND": SI_mean,
}

print(borough_means_dict)

## OR

borough_means = law33.groupby("BoroughName")["ENERGY STAR Score"].mean().to_dict()
print(borough_means)
```

```python
##Visualiztion

sorted_means = sorted(borough_means.items(), key=lambda x: x[1], reverse=True)

for borough, score in sorted_means:
    bar = "█" * int(score / 2)
    print(f"{borough:<15} | {bar} {score:.1f}")
```

```python
## The Hard Way
import csv

values = []

with open(
    "DOB_Sustainability_Compliance_Map__Local_Law_33_20261112.csv",
    mode="r",
    newline="",
    encoding="utf-8",
) as file:
    reader = csv.DictReader(file)
    for row in reader:
        if row["ENERGY STAR Score"].strip():
            values.append(float(row["ENERGY STAR Score"]))

mean_val = sum(values) / len(values)

sorted_values = sorted(values)
n = len(sorted_values)
mid = n // 2

if n % 2 == 0:
    median_val = (sorted_values[mid - 1] + sorted_values[mid]) / 2
else:
    median_val = sorted_values[mid]

counts = {}
for val in values:
    counts[val] = counts.get(val, 0) + 1

max_count = max(counts.values())
modes = [k for k, v in counts.items() if v == max_count]

print(f"Mean: {mean_val}")
print(f"Median: {median_val}")
print(f"Mode(s): {modes}")
```
